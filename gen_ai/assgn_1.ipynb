```python
import re
import pandas as pd
from collections import Counter

#TASK 2 & 7: ADVANCED PREPROCESSING & ERROR HANDLING
def preprocess_text(text):
    """
    Cleans raw text by removing numbers, URLs, repeated chars, 
    and short tokens while preserving context-heavy words.
    """
    # Handle empty strings or non-string inputs 
    if not text or not isinstance(text, str):
        return []

    # 1. Convert to lowercase
    text = text.lower()

    # 2. Remove URLs and Email patterns 
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)

    # 3. Remove numbers
    text = re.sub(r'\d+', '', text)

    # 4. Handle repeated characters 
    text = re.sub(r'(.)\1{2,}', r'\1', text)

    # 5. Remove special characters/punctuation 
    text = re.sub(r'[^\w\s]', '', text)

    # 6. Remove extra spaces [cite: 62]
    text = " ".join(text.split())

    # 7. Tokenization and filtering
    tokens = text.split()
  
    clean_tokens = [t for t in tokens if len(t) > 2 or t in ['no', 'not']]
    
    return clean_tokens

#TASK 6: FULL PIPELINE
def full_pipeline(text_list):
    """
    Processes a list of strings and returns structured data.
    """
    all_tokens = []
    clean_sentences = []
    
    for item in text_list:
        tokens = preprocess_text(item)
        all_tokens.append(tokens)
        clean_sentences.append(" ".join(tokens))
        
    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }

#TASK 3: STRESS TESTING
sample_inputs = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

# Running the pipeline
pipeline_output = full_pipeline(sample_inputs)

#TASK 4: TOKEN ANALYTICS
analysis_data = []
for i, original in enumerate(sample_inputs):
    tokens = pipeline_output["tokens"][i]
    
    total_tokens = len(tokens)
    unique_tokens = len(set(tokens))
    avg_len = sum(len(t) for t in tokens) / total_tokens if total_tokens > 0 else 0
    
    analysis_data.append({
        "Original Text": original,
        "Cleaned Tokens": tokens,
        "Cleaned Sentence": pipeline_output["clean_sentences"][i],
        "Total Tokens": total_tokens,
        "Unique Tokens": unique_tokens,
        "Avg Token Length": round(avg_len, 2)
    })

# Convert to DataFrame for clear display 
df_analytics = pd.DataFrame(analysis_data)
print("TASK 4: ANALYTICS TABLE")
print(df_analytics.to_string())

#TASK 5: FREQUENCY ANALYSIS
all_flattened_tokens = [token for sublist in pipeline_output["tokens"] for token in sublist]
word_counts = Counter(all_flattened_tokens)

print("\nTASK 5: FREQUENCY ANALYSIS")
print(f"Top 10 Most Frequent: {word_counts.most_common(10)}")
print(f"Top 5 Least Frequent: {word_counts.most_common()[:-6:-1]}")
```
