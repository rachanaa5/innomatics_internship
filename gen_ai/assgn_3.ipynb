```python

try:
    import torch
    from transformers import AutoModelForCausalLM, AutoTokenizer
    print("Libraries detected. Proceeding to chatbot logic...")
except ImportError:
    print("Installing required libraries (torch, transformers)...")
    %pip install torch transformers

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer


def load_chatbot_model(model_name="microsoft/DialoGPT-medium"):
    """
    Loads the pre-trained transformer model and tokenizer 
    from the Hugging Face Model Hub[cite: 239, 248].
    """
    print(f"Loading {model_name}. This may take a minute...")
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForCausalLM.from_pretrained(model_name)
    
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
        
    return tokenizer, model

def generate_response(user_input, tokenizer, model, chat_history_ids=None):
    """
    Generates a response using prompt-based text generation[cite: 254].
    Appends new input to history to maintain conversation flow[cite: 256].
    """
    
    new_user_input_ids = tokenizer.encode(user_input + tokenizer.eos_token, return_tensors='pt')

 
    if chat_history_ids is not None:
        bot_input_ids = torch.cat([chat_history_ids, new_user_input_ids], dim=-1)
    else:
        bot_input_ids = new_user_input_ids

    chat_history_ids = model.generate(
        bot_input_ids, 
        max_length=1000, 
        pad_token_id=tokenizer.eos_token_id,
        no_repeat_ngram_size=3,
        do_sample=True, 
        top_k=100, 
        top_p=0.7,
        temperature=0.8
    )

    response_ids = chat_history_ids[:, bot_input_ids.shape[-1]:]
    response_text = tokenizer.decode(response_ids[0], skip_special_tokens=True)
    
    return response_text, chat_history_ids

def run_chatbot():
    """
    Starts the console-based chatbot and handles the loop until exit[cite: 244, 278].
    """
    tokenizer, model = load_chatbot_model()
    chat_history_ids = None
    
    print("\n" + "="*50)
    print("Chatbot: Hello! I am your AI assistant. How can I help you today?")
    print("Type 'exit' or 'quit' to end the session [cite: 259-260].")
    print("="*50 + "\n")

    while True:
       
        user_input = input("User: ")

        if user_input.lower() in ['exit', 'quit']:
            print("Chatbot: Goodbye! Ending conversation[cite: 276].")
            break

        try:
            response, chat_history_ids = generate_response(
                user_input, tokenizer, model, chat_history_ids
            )
            print(f"Chatbot: {response}")
        except Exception as e:
            print(f"Chatbot: Oops! I ran into an error: {e}")

if __name__ == "__main__":
    run_chatbot()
```
